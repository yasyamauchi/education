```python
#インポート
import pandas as pd
import matplotlib.pyplot as plt

#irisデータセットの読み込み
from sklearn.datasets import load_breast_cancer
cancer_dataset = load_breast_cancer()
X_cancer = pd.DataFrame(cancer_dataset['data'], columns=cancer_dataset.feature_names) #説明変数 DataFrame型
y_cancer = cancer_dataset['target'] #目的変数

# 指定された特徴量のみを抽出
selected_cols = ['mean radius', 'mean texture', 'mean concave points']
X_cancer_selected = X_cancer[selected_cols]

# 抽出した特徴量で散布図行列を描画
plt.figure()
scatter_matrix_selected = pd.plotting.scatter_matrix(
    X_cancer_selected,
    c=y_cancer,  # 目的変数で色分け
    figsize=(8, 8),
    marker='o',
    hist_kwds={'bins': 20},
    edgecolors='black'
)
plt.suptitle('Scatter Matrix of Selected Mean Features', size=14)
plt.show()
```
