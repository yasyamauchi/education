## パーセプトロン

```python
# @title 単純パーセプトロンの例題(スライド)
x1 = 0.5 # @param {type:"number"}
x2 = 0.8 # @param {type:"number"}

w1 = 1 # @param {type:"slider", min:-5, max:5, step:1}
w2 = -2 # @param {type:"slider", min:-5, max:5, step:1}
phi = 0 # @param {type:"slider", min:-3, max:3, step:0.2}

print('x1*w1+x2*w2={}'.format(x1*w1+x2*w2))
if x1*w1+x2*w2 <= phi:
  print("y=0")
else:
  print("y=1")
```

```python
# @title 論理演算をパーセプトロンで実現(スライド)

x1 = True # @param ["False", "True"] {type:"raw"}
x2 = True # @param ["False", "True"] {type:"raw"}
w0 = -0.8 # @param {type:"slider", min:-2, max:2, step:0.1}
w1 = 0.5 # @param {type:"slider", min:-2, max:2, step:0.1}
w2 = 0.5 # @param {type:"slider", min:-2, max:2, step:0.1}

print('x0+x1w1+x2w2={}'.format(w0+w1*float(x1)+w2*float(x2)))
if w0+w1*float(x1)+w2*float(x2)<=0:
  y = False
else:
  y = True
print(y)
```

```python
# @title XORをパーセプトロンで実現(スライド)

x1 = True # @param ["False", "True"] {type:"raw"}
x2 = True # @param ["False", "True"] {type:"raw"}

def perceptron(x1,x2,w0,w1,w2):
  z = w0+w1*x1+w2*x2
  if z <= 0:
    return 0
  else:
    return 1

# ANDの実装（スライド通りの重みづけ）
def p_and(x1,x2):
  return perceptron(x1,x2,-0.8,0.5,0.5)

# NANDの実装（スライド通りの重みづけ）
def p_nand(x1,x2):
  return perceptron(x1,x2,0.8,-0.5,-0.5)

# ORの実装（スライド通りの重みづけ）
def p_or(x1,x2):
  return perceptron(x1,x2,-0.4,0.5,0.5)

y_nand = p_nand(x1,x2)
print('NANDの結果：{}'.format(bool(y_nand)))
y_or = p_or(x1,x2)
print('ORの結果：{}'.format(bool(y_or)))
y_xor = p_and(y_nand,y_or)
print('XORの結果：{}'.format(bool(y_xor)))
```

## 活性化関数

4つの活性化関数をグラフにする．

```python
from prompt_toolkit.shortcuts.dialogs import yes_no_dialog
import matplotlib.pyplot as plt
import numpy as np
import math

# ステップ関数 ※最小値は0としています
def step(x):
  return 1.0*(x>=0.0)

# シグモイド関数
def sigmoid(x):
  return 1 / (1 + e**-x)

# ReLU関数
def relu(x):
  return x*(x>=0.0)

# 恒等関数
def identity(x):
  return x

x = np.arange(-10, 10, 0.1)
e = math.e

ystep = step(x)
ysig = sigmoid(x)
yrelu = relu(x)
yident = identity(x)

fig = plt.figure()
ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)
ax4 = fig.add_subplot(2, 2, 4)
plt.subplots_adjust(hspace=0.5, wspace=0.4)
ax1.plot(x, ystep)
ax1.set_title('Step')
ax2.plot(x, ysig)
ax2.set_title('Sigmoid')
ax3.plot(x, yrelu)
ax3.set_title('ReLU')
ax4.plot(x, yident)
ax4.set_title('Identity')

plt.show()
```

##ニューラルネットワーク

###第0層→第1層  
次のベクトルと行列の計算を行う  
$$
x=\left(\begin{array}{}
    1.0 & 0.5
\end{array}\right)
$$

$$
b^{(1)}=\left(\begin{array}{}
    0.1 & 0.2 & 0.3
\end{array}\right)
$$

$$
W^{(1)}=\left(\begin{array}{}
    0.1 & 0.3 & 0.5 \\
    0.2 & 0.4 & 0.6
\end{array}\right)
$$

$$
a^{(1)}=xW^{(1)}+b^{(1)}=
\left(\begin{array}{}
    1.0 & 0.5
\end{array}\right)\left(\begin{array}{}
    0.1 & 0.3 & 0.5 \\
    0.2 & 0.4 & 0.6
\end{array}\right)+\left(\begin{array}{}
    0.1 & 0.2 & 0.3
\end{array}\right)
$$

$$
z^{(1)}=sigmoid(a^{(1)})
$$

```python
import numpy as np

# シグモイド関数
def sigmoid(x):
  return 1 / (1 + e**-x)

# x,b1,w1を代入する
x = np.array([1.0,0.5])
b1 = np.array([0.1, 0.2, 0.3])
w1 = np.array([[0.1,0.3,0.5],[0.2,0.4,0.6]])

# a1を計算する
a1=np.dot(x,w1)+b1   # np.dotは行列の積

# x,b1,w1,aを表示する
print('x={}'.format(x))
print('b1={}'.format(b1))
print('w1={}'.format(w1))
print('a1={}'.format(a1))

# a1を活性化関数に投入(活性化関数はシグモイド関数を利用)
z1 = sigmoid(a1)
print('Sigmoid(a1)={}'.format(z1))
```

###第1層→第2層    
※活性化関数はシグモイド関数  

$$
a^{(2)}=z^{(1)}W^{(2)}+b^{(2)}
$$

$$
z^{(2)}=sigmoid(a^{(2)})
$$



###第2層→第3層  
※活性化関数は恒等関数  

$$
a^{(3)}=z^{(2)}W^{(3)}+b^{(3)}
$$

$$
y=identity(a^{(3)})=a^{(3)}
$$
